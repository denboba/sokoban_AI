# 2. Heuristic Design Journal

## 2.1 Initial Approach

First Heuristic: Euclidean distance for box-target alignment.

Problem: Led to oscillations in easy_map1 (player stuck between two states).

Analysis:

Euclidean distance ignored obstacles, causing unrealistic paths.

Switched to Manhattan distance + path accessibility checks (path_exists()).

Second Iteration: Added deadlock detection.

Trigger: In hard_map1, boxes got stuck in corners.

Solution:

deadlock_heuristic() penalizes:

Boxes in corners (adjacent_obstacles >= 2).

Box pairs blocking each other (horizontal_blocked/vertical_blocked).

Impact: Reduced invalid states by 40%.

Final Heuristic: Hungarian Algorithm for optimal box-target matching.

Optimization:

Combined with player-to-box distance to prioritize pushing.

Result: 0-pull solutions for easy/medium maps.
